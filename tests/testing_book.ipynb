```python
from dbx import query_df

# Smoke test
print(query_df("SELECT 1 AS ok"))

# Read a few rows
df = query_df("SELECT * FROM veiturdata_base_prd.ami.smartmeter LIMIT 5")
print(df.head())

# Your hourly demand test (pure SQL)
sql_text = """
SELECT
  hour(`timestamp`) AS hr,
  CASE resulttype
    WHEN 'AMP_Power_L1_AVG_T0' THEN 'L1'
    WHEN 'AMP_Power_L2_AVG_T0' THEN 'L2'
    WHEN 'AMP_Power_L3_AVG_T0' THEN 'L3'
  END AS phase,
  SUM(CAST(value AS DOUBLE)) AS p_sum
FROM veiturdata_base_prd.ami.smartmeter
WHERE resulttype IN (:l1, :l2, :l3)
  AND `timestamp` >= date_sub(current_timestamp(), :days_back)
GROUP BY hour(`timestamp`), resulttype
ORDER BY hr;
"""
df = query_df(sql_text, {
    "l1":"AMP_Power_L1_AVG_T0",
    "l2":"AMP_Power_L2_AVG_T0",
    "l3":"AMP_Power_L3_AVG_T0",
    "days_back": 30
})
print(df.head())
```
